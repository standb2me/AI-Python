```python
def asmd(a, b, op):
    if op == "+":
        return a + b
    elif op == "-":
        return a - b
    elif op == "*":
        return a * b
    elif op == "/":
        if b == 0:
            return "0으로 나눌 수 없습니다."
        return a / b
    else:
        return "잘못된 연산자입니다."


x = int(input("첫 번째 수: "))
y = int(input("두 번째 수: "))
op = input("연산자 입력 (+, -, *, /): ")

result = asmd(x, y, op)
print(f"{x} {op} {y} = {result}")
```

```python
def asmd(a, b, op):
    if op == "+":
        return f"{a} + {b} = {a + b}"
    elif op == "-":
        return f"{a} - {b} = {a - b}"
    elif op == "*":
        return f"{a} * {b} = {a * b}"
    elif op == "/":
        if b == 0:
            return "0으로 나눌 수 없습니다."
        return f"{a} / {b} = {a / b:.2f}"
    else:
        return "잘못된 연산자입니다."


x = int(input("첫 번째 수: "))
y = int(input("두 번째 수: "))
op = input("연산자 입력 (+, -, *, /): ")

result = asmd(x, y, op)
print(result)
```

```python
import random

com = random.randint(1, 100)
print(com)  

for i in range(5):
    guess = int(input(f"{i+1}번째 시도 - 1~100 중 하나를 맞혀 보세요: "))

    if guess == com:
        print("당신의 숫자는", com, "와 같습니다. 정답입니다!")
        break
    elif guess > com:
        print("당신의 숫자는", com, "보다 큽니다.")
    else:
        print("당신의 숫자는", com, "보다 작습니다.")
else:
  
    print(f"아쉽습니다! 5번의 기회를 모두 다 쓰셨습니다. 정답은 {com}입니다.")
```

```python
import random

com = random.randint(1, 100)
count = 0  

while True:
    guess = int(input("1~100 중 하나를 맞혀 보세요: "))
    print(com)
    count += 1  

    if guess == com:
        print(f"정답입니다! {count}번 만에 맞혔습니다.")
        break  
    elif guess > com:
        print("입력한 숫자가 정답보다 큽니다.")
    else:
        print("입력한 숫자가 정답보다 작습니다.")
```
